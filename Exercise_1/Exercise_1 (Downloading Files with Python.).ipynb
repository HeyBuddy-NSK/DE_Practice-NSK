```python
download_uris = [
    "https://divvy-tripdata.s3.amazonaws.com/Divvy_Trips_2018_Q4.zip",
    "https://divvy-tripdata.s3.amazonaws.com/Divvy_Trips_2019_Q1.zip",
    "https://divvy-tripdata.s3.amazonaws.com/Divvy_Trips_2019_Q2.zip",
    "https://divvy-tripdata.s3.amazonaws.com/Divvy_Trips_2019_Q3.zip",
    "https://divvy-tripdata.s3.amazonaws.com/Divvy_Trips_2019_Q4.zip",
    "https://divvy-tripdata.s3.amazonaws.com/Divvy_Trips_2020_Q1.zip",
    "https://divvy-tripdata.s3.amazonaws.com/Divvy_Trips_2220_Q1.zip",
]
```

```python
def download_file(url: str, save_path: str) -> None:
    """Download a single file from the given URL.

    Parameters
    ----------

    :url : URL of the file to download.
    :save_path : Path where the downloaded file will be saved.
    """
    print("Starting download!.")

    # Using context manager with requests.
    with requests.get(url, stream=True, timeout=7) as resp:

        # raising error if any for response status.
        resp.raise_for_status()
        
        with open(save_path, 'wb') as file:

            # using iter_content() function to save the file by small alive chunks.
            for content in resp.iter_content(chunk_size=8196):
                if content:
                    file.write(content)
                    
    print("File download Complete!.")
    
```

```python
download_file(download_uris[0],'second_s_file.zip')
```

```python
def check_valid_url(url: str) -> bool:
    try:
        parsed_url = urlparse(url)
        print(parsed_url)
        return parsed_url.scheme in ('http','https') or parsed_url.netloc
    except Exception as e:
        return False
```

```python
for url in download_uris:
    print(check_valid_url(url))
```
